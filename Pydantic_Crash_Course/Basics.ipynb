# **Pydantic** is used mostly for Data Validataion

```python
from pydantic import BaseModel,Field
class User(BaseModel):
    name:str
    age:int
```

```python
user1=User(name="Fairoz",age=20)
```

```python
user1.name, user1.age
```

# **Field Validation**

```python
class Voter(BaseModel):
    nationality:str=Field(max_length=10,default="Indian")
    age:int=Field(ge=18)
    city:str=Field(min_length=5,default="Not Entered")
```

```python
v1=Voter(nationality="indian",age=21,city="Srinagar")
v2=Voter(nationality="indian",age=21)
```

```python
v1
```

```python
v2
```

# **Lists**

```python
from typing import List
class Product(BaseModel):
    name:str=Field(min_length=5)
    description:str=Field(min_length=6)
    price:int=Field(ge=0)

class Cart(BaseModel):
    id:str=Field(min_length=5,max_length=10)
    products:List[Product]
```

```python
prod1=Product(name="Samsung",description="Mobile Phone",price=21)
prod2=Product(name="13 Pro Max",description="IPhone",price=32)
cart=Cart(id="ab1c2n_32",products=[prod1,prod2])

print(cart)
```

```python
cart.id
```

```python
cart.products
```

```python
cart.products[0].name
```

# **Annotated** as a Common Validator

```python
from typing import Annotated

common_validator=Annotated[List[Product],Field(default=[])]
```

```python
class Carts(BaseModel):
    id:str=Field(min_length=2,max_length=5)
    items:common_validator
    comments:str=Field(default="No Comments")
class Order(BaseModel):
    products:common_validator

order=Order()
```

```python
order
```

```python
ord1=Order(products=[prod1,prod2])
```
