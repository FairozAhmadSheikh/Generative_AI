```python
from langchain_groq import ChatGroq
llm = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0
)
```

```python
text="Hello My name is fairoz ahmad Sheikh and my email is fairozahmadsheikh@gmail.com my age is 100 years"
```

```python
res=llm.invoke(f'Extract name email age only from the text {text}')
print(res.content)
```

```python
type(res.content)
```

# **Structured Outputs**

```python
from pydantic import BaseModel,Field
from typing import List

class ResponseModel(BaseModel):
    name:str=Field(description="Your Name")
    email:str=Field(description="Email Address")
    age:int=Field(description='Age ')
    
```

```python
structured_llm=llm.with_structured_output(ResponseModel)
res=structured_llm.invoke(f'Extract name email age only from the text {text}')
print(res)
```

```python
print(res.name,type(res.name))
```

```python
res.age,type(res.age)
```

```python
res.email
```

# **A bit Advanced**

```python
from pydantic import BaseModel, Field
from typing import List

class Movies(BaseModel):
    title:str=Field(description="Title of the Movie")
    year:int=Field(description="Movie year")

class All_Movies(BaseModel):
    movies:List[Movies]=Field(description="List of Movies")


```

```python
single_movie_llm=llm.with_structured_output(Movies)
res=single_movie_llm.invoke("Provide Sci-fi movies")
print(res)
```

```python
multiple_movies_list=llm.with_structured_output(All_Movies)
resp=multiple_movies_list.invoke('Provide a list of Scifi Movies')
print(resp)
```
