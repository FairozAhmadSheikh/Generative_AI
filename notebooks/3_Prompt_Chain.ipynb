```python
from langchain_groq import ChatGroq

llm = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0
)
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
prompts=ChatPromptTemplate.from_messages([
    {"role":"system","content":"You are a language translator. Translate input in {language}"},
    {"role":"user","content":"{query}"}
])
prompts
```

```python
final_prompt=prompts.format_messages(language="Urdu",query="I love people with good communication skills")
final_prompt
```

```python
res=llm.invoke(final_prompt)
print(res.content)
```

# **Chains** -> Runnables

```python
chains=prompts | llm
res=chains.invoke({"language":"Urdu","query":"I Love playing"})
print(res.content)
```

```python
# Output Parser

out=StrOutputParser()
```

# **Custom Chain**


```python
def transform_response(result:str):
    return result.upper()
```

```python
chains=prompts | llm | out |transform_response
```

```python
res=chains.invoke({"language":"spanish","query":"Arigato"})
print(res)
```
