```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.tools import tool
from langchain.agents import create_agent
```

```python
@tool
def add_numbers(a:int,b:int):
    """
    This tool will be used to add two numbers

    Args:
    a : Number First
    b : Number Second
    """
    return a+b
@tool

def multiply_numbers(a:int,b:int):
    """
    This tool will be used to multiply two numbers

     Args:
        a : Number First
        b : Number Second
    """
    return a * b  
```

```python
from langchain_groq import ChatGroq
llm = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0
)
```

```python
agent=create_agent(
    model=llm,
    tools=[add_numbers,multiply_numbers],
    system_prompt="You are a Math Teacher. Use tools to perform Calculations"

)

response=agent.invoke({"messages":[{'role':'user',"content":"What is 5+7 *100"}]})

for res in response['messages']:
    print(res)
```
