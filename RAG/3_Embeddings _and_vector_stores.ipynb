```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_ollama import OllamaEmbeddings,ChatOllama
from langchain_chroma import Chroma
```

```python
loader=PyPDFLoader('../Data/DT_Gene_Fairoz.pdf')
data=loader.load()
data
```

```python
llm=ChatOllama(
    model="openai/gpt-oss-20b"
)
```

# **Embedding Creation**

```python
splitter=RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=100)
chunks=splitter.split_documents(documents=data)

texts=[chunk.page_content for chunk in chunks]
```

```python
embeddings=OllamaEmbeddings(model="nomic-embed-text")
vectors=embeddings.embed_documents(texts)
print(len(vectors))
```

# **Embedding Store**

```python
vector_store=Chroma.from_texts(texts=texts,embedding=embeddings,persist_directory="./vector_db")
```

```python
query="something about experience"

result=vector_store.similarity_search(query=query,k=2)
print(result)
```
